```python
# Functions for training and evaluation

class AverageMeter:
    def __init__(self):
        self.reset()

    def reset(self):
        self.val = 0
        self.avg = 0
        self.sum = 0
        self.count = 0

    def update(self, val, n=1):
        self.val = val
        self.sum += val * n
        self.count += n
        self.avg = self.sum / self.count

def train(model, train_loader, criterion, optimizer, device):
    model.train()
    
    train_loss_avg = AverageMeter()
    train_acc_avg = AverageMeter()

    for input, label in tqdm(train_loader, desc="Training", leave=False):
        input, label = input.to(device), label.to(device)

        # Forward
        optimizer.zero_grad()
        outputs = model(input)
        loss = criterion(outputs, label)

        # Backward
        loss.backward()
        optimizer.step()
        
        # Predictions 
        _, preds = torch.max(outputs, 1)
        
        # Metrics for batch
        batch_acc = (preds == label).sum().item() / label.size(0)

        train_loss_avg.update(loss.item(), label.size(0))
        train_acc_avg.update(batch_acc, label.size(0))

    return train_loss_avg.avg, train_acc_avg.avg

def evaluate(model, val_loader, criterion, device):
    model.eval()
    
    val_loss_avg = AverageMeter()
    val_acc_avg = AverageMeter()
    
    with torch.no_grad():
        for input, label in tqdm(val_loader, desc="Validation", leave=False):
            input, label = input.to(device), label.to(device)

            # Forward
            outputs = model(input)
            loss = criterion(outputs, label)

            # Predictions
            _, preds = torch.max(outputs, 1)

            # Metrics for batch
            batch_acc = (preds == label).sum().item() / label.size(0)

            val_loss_avg.update(loss.item(), label.size(0))
            val_acc_avg.update(batch_acc, label.size(0))

    return val_loss_avg.avg, val_acc_avg.avg

def create_mobilenetv2_model(description_path=None, config=None):
    # create model instance
    model = models.mobilenet_v2(weights=models.MobileNet_V2_Weights.IMAGENET1K_V1)
    model.classifier[1] = nn.Linear(in_features=1280, out_features=config['num_classes'])
    model = model.to(config['device'])
    
    # count parameters
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

    # save model description
    if description_path:
        os.makedirs(os.path.dirname(description_path), exist_ok=True)
        with open(description_path, "w") as f:
            f.write(f"Total parameters: {total_params}\n")
            f.write(f"Trainable parameters: {trainable_params}\n")
            f.write(f"Model Architecture:\n{model}\n")
    
    # Loss , Optimizer and Scheduler
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=config['lr'])
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.1)
    
    return model, criterion, optimizer, scheduler
```

```python
# calcular o número de flops do modelo
def count_flops(model, input_size=(1, 3, 200, 200)):
    from thop import profile
    dummy_input = torch.randn(input_size).to(next(model.parameters()).device)
    flops, params = profile(model, inputs=(dummy_input,), verbose=False)
    return flops, params
```

```python
# Configuration for Larvae Dataset

CONFIG_EGG = {
    'dataset_name': 'eggs',
    'split': [1, 2, 3],
    'percentage': [1, 5, 25, 50, 75, 100],
    'num_classes': 9,
    'batch_size': 16,
    'num_workers': 4,
    'image_size': 200,
    'lr': 1e-4,
    'num_epochs': 100,
    'device': 'cuda' if torch.cuda.is_available() else 'cpu'
}

# Create dataloaders
dataloaders_egg = {}

transforms_egg = transforms.Compose([
    transforms.Resize((CONFIG_EGG['image_size'], CONFIG_EGG['image_size'])),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

for split in CONFIG_EGG['split']:
    dataloaders_egg[split] = {}
    for perc in CONFIG_EGG['percentage']:
        data_module = DataModuleParasite(
            dataset_name=CONFIG_EGG['dataset_name'],
            split=split,
            percentage=perc,
            batch_size=CONFIG_EGG['batch_size'],
            num_workers=CONFIG_EGG['num_workers'],
            transform=transforms_egg
        )
        data_module.setup()
        dataloaders_egg[split][perc] = {
            'train': data_module.train_dataloader(),
            'val': data_module.val_dataloader(),
            'test': data_module.test_dataloader()
        }

mobilenetv2_model_pretrained, _, _, _ = create_mobilenetv2_model(description_path="mobilenetv2/eggs/mobilenetv2_model_params_eggs.txt", config=CONFIG_EGG)
```

```python
n_flops, n_params = count_flops(mobilenetv2_model_pretrained, input_size=(1, 3, CONFIG_EGG['image_size'], CONFIG_EGG['image_size']))
print(f"Model: mobilenetv2 - Dataset: {CONFIG_EGG['dataset_name']}")
# mostra o valor total
print(f"FLOPs: {n_flops} FLOPs")
print(f"Parameters: {n_params} ")
```

```python
historic_eggs = {}
patience = 20 

for perc in CONFIG_EGG['percentage']:
    historic_eggs[perc] = {}
    for split in CONFIG_EGG['split']:
        
        mobilenetv2_model_pretrained, criterion, optimizer, scheduler = create_mobilenetv2_model(None, CONFIG_EGG)
        
        epochs_no_improve = 0
        
        best_val_acc = 0.0
        historic_eggs[perc][split] = {
            'train_loss': [],
            'train_acc': [],
            'val_loss': [],
            'val_acc': []
        }
        
        for epoch in tqdm(range(CONFIG_EGG['num_epochs']), desc=f"Split {split} - Percentage {perc}%", leave=False):
            
            val_loss, val_acc = evaluate(mobilenetv2_model_pretrained, dataloaders_egg[split][perc]['val'], criterion, CONFIG_EGG['device'])
            
            train_loss, train_acc = train(mobilenetv2_model_pretrained, dataloaders_egg[split][perc]['train'], criterion, optimizer, CONFIG_EGG['device'])
                
            historic_eggs[perc][split]['train_loss'].append(train_loss)
            historic_eggs[perc][split]['train_acc'].append(train_acc)
            historic_eggs[perc][split]['val_loss'].append(val_loss)
            historic_eggs[perc][split]['val_acc'].append(val_acc)
            
            # Early Stopping
            if val_acc > best_val_acc:
                best_val_acc = val_acc
                epochs_no_improve = 0
                os.makedirs('mobilenetv2/eggs/best_models_eggs', exist_ok=True)
                torch.save(mobilenetv2_model_pretrained.state_dict(), f'mobilenetv2/eggs/best_models_eggs/mobilenetv2_best_split{split}_perc{perc}.pth')
            else:
                epochs_no_improve += 1
                if epochs_no_improve >= patience:
                    print(f'Early stopping at epoch {epoch+1} for split {split} and percentage {perc}%')
                    break
                
        val_loss, val_acc = evaluate(mobilenetv2_model_pretrained, dataloaders_egg[split][perc]['val'], criterion, CONFIG_EGG['device'])
        
        historic_eggs[perc][split]['val_loss'].append(val_loss)
        historic_eggs[perc][split]['val_acc'].append(val_acc)
            
        scheduler.step()
        
# Save training history
with open('mobilenetv2/eggs/mobilenetv2_historic_eggs.json', 'w') as f:
    json.dump(historic_eggs, f) 
```

```python
# Evaluate on test set and save metrics
results_eggs = {}

for perc in CONFIG_EGG['percentage']:
    results_eggs[perc] = {}
    for split in CONFIG_EGG['split']:
        
        # Load the best model
        mobilenetv2_model_pretrained.load_state_dict(torch.load(f'mobilenetv2/eggs/best_models_eggs/mobilenetv2_best_split{split}_perc{perc}.pth'))
        
        mobilenetv2_model_pretrained.eval()
        
        all_preds = []
        all_labels = []
        
        with torch.no_grad():
            for inputs, labels in tqdm(dataloaders_egg[split][perc]['test'], desc=f'Testing Split {split} Perc {perc}%', leave=False):
                inputs, labels = inputs.to(CONFIG_EGG['device']), labels.to(CONFIG_EGG['device'])
                outputs = mobilenetv2_model_pretrained(inputs)
                _, preds = torch.max(outputs, 1)
                
                all_preds.extend(preds.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())
        
        # Calculate metrics
        cm = confusion_matrix(all_labels, all_preds)
        kappa = cohen_kappa_score(all_labels, all_preds)
        report_dic = classification_report(all_labels, all_preds, zero_division=0, output_dict=True)
        report_str = classification_report(all_labels, all_preds, zero_division=0) 
        
        # Store results
        results_eggs[perc][split] = {
            'confusion_matrix': cm.tolist(),
            'kappa': float(kappa),
            'classification_report_dict': report_dic,
            'classification_report_str': report_str
        }
        
# Save results to JSON
with open('mobilenetv2/eggs/mobilenetv2_test_results_eggs.json', 'w') as f:
    json.dump(results_eggs, f)
```

```python
load = False
# Load results from JSON - commented out because already loaded above
# load = True
# with open('mobilenetv2/eggs/mobilenetv2_test_results_eggs.json', 'r') as f:
#     results_eggs = json.load(f)

# delete any existing aggregated report file
if os.path.exists('mobilenetv2/eggs/mobilenetv2_aggregated_classification_report_eggs.txt'):
    os.remove('mobilenetv2/eggs/mobilenetv2_aggregated_classification_report_eggs.txt')

# Function to generate mean ± std report including Cohen Kappa
def generate_mean_std_report(results_data, percentage, split_list):
    reports_list = []
    kappa_list = [] 

    # 1. Collect the data (Reports and Kappas)
    for split in split_list:
        data_split = results_data[percentage][split]
        if 'classification_report_dict' in data_split:
            reports_list.append(data_split['classification_report_dict'])
            kappa_list.append(data_split['kappa'])
        else:
            return None

    # 2. Prepare the DataFrame
    keys = list(reports_list[0].keys())
    # Remove 'accuracy' from keys if it appears as a standalone key (handled separately)
    if 'accuracy' in keys: keys.remove('accuracy')
    
    final_df = pd.DataFrame(index=keys + ['accuracy', 'Cohen Kappa'], 
                            columns=['precision', 'recall', 'f1-score', 'support'])

    # 3. Fill Classes and Averages (Macro/Weighted)
    for key in keys:
        for metric in ['precision', 'recall', 'f1-score', 'support']:
            values = [r[key][metric] for r in reports_list]
            mean_val, std_val = np.mean(values), np.std(values)
            
            if metric == 'support':
                final_df.loc[key, metric] = f"{int(mean_val)}"
            else:
                final_df.loc[key, metric] = f"{mean_val:.4f} ± {std_val:.4f}"

    # 4. Fill Accuracy
    acc_values = [r['accuracy'] for r in reports_list]
    final_df.loc['accuracy', 'f1-score'] = f"{np.mean(acc_values):.4f} ± {np.std(acc_values):.4f}"
    # Optional: put total support in the accuracy row
    total_support = int(np.mean([r['macro avg']['support'] for r in reports_list]))
    final_df.loc['accuracy', 'support'] = f"{total_support}"

    # 5. Fill Cohen Kappa (NEW)
    # Kappa doesn't have precision/recall, so we put the result in the f1-score column (or create a new one)
    k_mean = np.mean(kappa_list)
    k_std = np.std(kappa_list)
    final_df.loc['Cohen Kappa', 'f1-score'] = f"{k_mean:.4f} ± {k_std:.4f}"

    # Replace NaN with empty string for cleaner visual appearance
    return final_df.fillna('')

print("=== Aggregated Reports (Mean ± Standard Deviation) ===")

for perc in CONFIG_EGG['percentage']:
    # print(f"\n>> Percentage: {perc}%")
    df_result = generate_mean_std_report(results_eggs, perc, CONFIG_EGG['split'])
    
    if df_result is not None:
        # Cosmetic adjustment to match sklearn print
        # Replace NaN with empty string to clean the accuracy line
        # print(df_result.fillna('').to_markdown()) 
        
        # Optional: Save all percentages in a single txt        
        with open('mobilenetv2/eggs/mobilenetv2_aggregated_classification_report_eggs.txt', 'a') as f:
            f.write(f"\n>> Percentage: {perc}%\n")
            f.write(df_result.fillna('').to_string())
            f.write("\n\n")
```

```python
# Visualization settings
sns.set_style("whitegrid")
plt.rcParams.update({'font.size': 12})

def normalize_cm(cm):
    """
    Normalize a confusion matrix by row (true labels).

    """
    cm = np.array(cm)
    with np.errstate(divide='ignore', invalid='ignore'):
        cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
    return np.nan_to_num(cm_norm) # Replace NaNs with 0

# ==============================================================================
# 1. FULL GRID: Rows = Percentages, Columns = Splits
# ==============================================================================

percentages = CONFIG_EGG['percentage']
splits = CONFIG_EGG['split']

n_rows = len(percentages)
n_cols = len(splits)

# Create the figure (adjustable size: width x height in inches)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(5 * n_cols, 4 * n_rows))

# If there is only 1 row or 1 column, ensure axes is a 2D matrix
if n_rows == 1: axes = np.expand_dims(axes, axis=0)
if n_cols == 1: axes = np.expand_dims(axes, axis=1)

print("Generating Grid of Matrices (Percentage x Split)...")
for i, perc in enumerate(percentages):
    for j, split in enumerate(splits):
        ax = axes[i, j]
        
        # Retrieve and normalize
        cm_raw = results_eggs[perc][split]['confusion_matrix']
        cm_norm = normalize_cm(cm_raw)
        
        # Plot Heatmap
        sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Blues', 
                    cbar=False, ax=ax, vmin=0, vmax=1, annot_kws={"size": 10})
        
        # Titles and Labels
        if i == 0:
            ax.set_title(f'Split {split}', fontsize=14, fontweight='bold')
        
        if j == 0:
            ax.set_ylabel(f'Perc {perc}%\nTrue Label', fontsize=12, fontweight='bold')
        else:
            ax.set_ylabel('') # Remove internal label to clean visual
            
        if i == n_rows - 1:
            ax.set_xlabel('Predicted Label', fontsize=12)
        else:
            ax.set_xlabel('')

plt.tight_layout()
plt.savefig('mobilenetv2/eggs/confusion_matrix_grid_full.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# ==============================================================================
# 2. AGGREGATED MATRIX: One matrix per Percentage (Sum of splits -> Normalization)
# ==============================================================================

print("Generating Aggregated Matrices (Average of Splits per Percentage)...")

# Create figure for aggregated matrices (one row, multiple columns)
fig_agg, axes_agg = plt.subplots(1, len(percentages), figsize=(5 * len(percentages), 5))

# Ensure iterability if there is only 1 percentage
if len(percentages) == 1: axes_agg = [axes_agg]

for i, perc in enumerate(percentages):
    ax = axes_agg[i]
    
    # 1. Sum the raw matrices of all splits for this percentage
    # This is better than averaging the normalized ones to handle imbalanced classes
    sum_cm = None
    for split in splits:
        cm_current = np.array(results_eggs[perc][split]['confusion_matrix'])
        if sum_cm is None:
            sum_cm = cm_current
        else:
            sum_cm += cm_current
            
    # 2. Normalize the sum
    cm_final_norm = normalize_cm(sum_cm)
    
    # Plot
    sns.heatmap(cm_final_norm, annot=True, fmt='.2f', cmap='Greens', 
                cbar=False, ax=ax, vmin=0, vmax=1, square=True)
    
    ax.set_title(f'Percentage {perc}% (Aggregated)', fontsize=13, fontweight='bold')
    ax.set_ylabel('True Label')
    ax.set_xlabel('Predicted Label')

plt.tight_layout()
plt.savefig('mobilenetv2/eggs/confusion_matrix_aggregated.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
historic_cistos = {}
patience = 20 

for perc in CONFIG_CISTO['percentage']:
    historic_cistos[perc] = {}
    for split in CONFIG_CISTO['split']:
        
        mobilenetv2_model_pretrained, criterion, optimizer, scheduler = create_mobilenetv2_model(None, CONFIG_CISTO)
        
        epochs_no_improve = 0
        
        best_val_acc = 0.0
        historic_cistos[perc][split] = {
            'train_loss': [],
            'train_acc': [],
            'val_loss': [],
            'val_acc': []
        }
        
        for epoch in tqdm(range(CONFIG_CISTO['num_epochs']), desc=f"Split {split} - Percentage {perc}%", leave=False):
            
            val_loss, val_acc = evaluate(mobilenetv2_model_pretrained, dataloaders_cisto[split][perc]['val'], criterion, CONFIG_CISTO['device'])
            
            train_loss, train_acc = train(mobilenetv2_model_pretrained, dataloaders_cisto[split][perc]['train'], criterion, optimizer, CONFIG_CISTO['device'])
                
            historic_cistos[perc][split]['train_loss'].append(train_loss)
            historic_cistos[perc][split]['train_acc'].append(train_acc)
            historic_cistos[perc][split]['val_loss'].append(val_loss)
            historic_cistos[perc][split]['val_acc'].append(val_acc)
            
            # Early Stopping
            if val_acc > best_val_acc:
                best_val_acc = val_acc
                epochs_no_improve = 0
                os.makedirs('mobilenetv2/cistos/best_models_cistos', exist_ok=True)
                torch.save(mobilenetv2_model_pretrained.state_dict(), f'mobilenetv2/cistos/best_models_cistos/mobilenetv2_best_split{split}_perc{perc}.pth')
            else:
                epochs_no_improve += 1
                if epochs_no_improve >= patience:
                    print(f'Early stopping at epoch {epoch+1} for split {split} and percentage {perc}%')
                    break
                
        val_loss, val_acc = evaluate(mobilenetv2_model_pretrained, dataloaders_cisto[split][perc]['val'], criterion, CONFIG_CISTO['device'])
        
        historic_cistos[perc][split]['val_loss'].append(val_loss)
        historic_cistos[perc][split]['val_acc'].append(val_acc)
            
        scheduler.step()
        
# Save training history
with open('mobilenetv2/cistos/mobilenetv2_historic_cistos.json', 'w') as f:
    json.dump(historic_cistos, f) 
```

```python
load = False
# Load results from JSON - commented out because already loaded above
# load = True
# with open('mobilenetv2/cistos/mobilenetv2_test_results_cistos.json', 'r') as f:
#     results_cistos = json.load(f)

# delete any existing aggregated report file
if os.path.exists('mobilenetv2/cistos/mobilenetv2_aggregated_classification_report_cistos.txt'):
    os.remove('mobilenetv2/cistos/mobilenetv2_aggregated_classification_report_cistos.txt')

# Function to generate mean ± std report including Cohen Kappa
def generate_mean_std_report(results_data, percentage, split_list):
    reports_list = []
    kappa_list = [] 

    # 1. Collect the data (Reports and Kappas)
    for split in split_list:
        data_split = results_data[percentage][split]
        if 'classification_report_dict' in data_split:
            reports_list.append(data_split['classification_report_dict'])
            kappa_list.append(data_split['kappa'])
        else:
            return None

    # 2. Prepare the DataFrame
    keys = list(reports_list[0].keys())
    # Remove 'accuracy' from keys if it appears as a standalone key (handled separately)
    if 'accuracy' in keys: keys.remove('accuracy')
    
    final_df = pd.DataFrame(index=keys + ['accuracy', 'Cohen Kappa'], 
                            columns=['precision', 'recall', 'f1-score', 'support'])

    # 3. Fill Classes and Averages (Macro/Weighted)
    for key in keys:
        for metric in ['precision', 'recall', 'f1-score', 'support']:
            values = [r[key][metric] for r in reports_list]
            mean_val, std_val = np.mean(values), np.std(values)
            
            if metric == 'support':
                final_df.loc[key, metric] = f"{int(mean_val)}"
            else:
                final_df.loc[key, metric] = f"{mean_val:.4f} ± {std_val:.4f}"

    # 4. Fill Accuracy
    acc_values = [r['accuracy'] for r in reports_list]
    final_df.loc['accuracy', 'f1-score'] = f"{np.mean(acc_values):.4f} ± {np.std(acc_values):.4f}"
    # Optional: put total support in the accuracy row
    total_support = int(np.mean([r['macro avg']['support'] for r in reports_list]))
    final_df.loc['accuracy', 'support'] = f"{total_support}"

    # 5. Fill Cohen Kappa (NEW)
    # Kappa doesn't have precision/recall, so we put the result in the f1-score column (or create a new one)
    k_mean = np.mean(kappa_list)
    k_std = np.std(kappa_list)
    final_df.loc['Cohen Kappa', 'f1-score'] = f"{k_mean:.4f} ± {k_std:.4f}"

    # Replace NaN with empty string for cleaner visual appearance
    return final_df.fillna('')

print("=== Aggregated Reports (Mean ± Standard Deviation) ===")

for perc in CONFIG_CISTO['percentage']:
    # print(f"\n>> Percentage: {perc}%")
    df_result = generate_mean_std_report(results_cistos, perc, CONFIG_CISTO['split'])
    
    if df_result is not None:
        # Cosmetic adjustment to match sklearn print
        # Replace NaN with empty string to clean the accuracy line
        # print(df_result.fillna('').to_markdown()) 
        
        # Optional: Save all percentages in a single txt        
        with open('mobilenetv2/cistos/mobilenetv2_aggregated_classification_report_cistos.txt', 'a') as f:
            f.write(f"\n>> Percentage: {perc}%\n")
            f.write(df_result.fillna('').to_string())
            f.write("\n\n")
```

```python
load = False
# Load results from JSON - commented out because already loaded above
# load = True
# with open('mobilenetv2/larvae/mobilenetv2_test_results_larvae.json', 'r') as f:
#     results_larvae = json.load(f)

# delete any existing aggregated report file
if os.path.exists('mobilenetv2/larvae/mobilenetv2_aggregated_classification_report_larvae.txt'):
    os.remove('mobilenetv2/larvae/mobilenetv2_aggregated_classification_report_larvae.txt')

# Function to generate mean ± std report including Cohen Kappa
def generate_mean_std_report(results_data, percentage, split_list):
    reports_list = []
    kappa_list = [] 

    # 1. Collect the data (Reports and Kappas)
    for split in split_list:
        data_split = results_data[percentage][split]
        if 'classification_report_dict' in data_split:
            reports_list.append(data_split['classification_report_dict'])
            kappa_list.append(data_split['kappa'])
        else:
            return None

    # 2. Prepare the DataFrame
    keys = list(reports_list[0].keys())
    # Remove 'accuracy' from keys if it appears as a standalone key (handled separately)
    if 'accuracy' in keys: keys.remove('accuracy')
    
    final_df = pd.DataFrame(index=keys + ['accuracy', 'Cohen Kappa'], 
                            columns=['precision', 'recall', 'f1-score', 'support'])

    # 3. Fill Classes and Averages (Macro/Weighted)
    for key in keys:
        for metric in ['precision', 'recall', 'f1-score', 'support']:
            values = [r[key][metric] for r in reports_list]
            mean_val, std_val = np.mean(values), np.std(values)
            
            if metric == 'support':
                final_df.loc[key, metric] = f"{int(mean_val)}"
            else:
                final_df.loc[key, metric] = f"{mean_val:.4f} ± {std_val:.4f}"

    # 4. Fill Accuracy
    acc_values = [r['accuracy'] for r in reports_list]
    final_df.loc['accuracy', 'f1-score'] = f"{np.mean(acc_values):.4f} ± {np.std(acc_values):.4f}"
    # Optional: put total support in the accuracy row
    total_support = int(np.mean([r['macro avg']['support'] for r in reports_list]))
    final_df.loc['accuracy', 'support'] = f"{total_support}"

    # 5. Fill Cohen Kappa (NEW)
    # Kappa doesn't have precision/recall, so we put the result in the f1-score column (or create a new one)
    k_mean = np.mean(kappa_list)
    k_std = np.std(kappa_list)
    final_df.loc['Cohen Kappa', 'f1-score'] = f"{k_mean:.4f} ± {k_std:.4f}"

    # Replace NaN with empty string for cleaner visual appearance
    return final_df.fillna('')

print("=== Aggregated Reports (Mean ± Standard Deviation) ===")

for perc in CONFIG_LARVAE['percentage']:
    # print(f"\n>> Percentage: {perc}%")
    df_result = generate_mean_std_report(results_larvae, perc, CONFIG_LARVAE['split'])
    
    if df_result is not None:
        # Cosmetic adjustment to match sklearn print
        # Replace NaN with empty string to clean the accuracy line
        # print(df_result.fillna('').to_markdown()) 
        
        # Optional: Save all percentages in a single txt        
        with open('mobilenetv2/larvae/mobilenetv2_aggregated_classification_report_larvae.txt', 'a') as f:
            f.write(f"\n>> Percentage: {perc}%\n")
            f.write(df_result.fillna('').to_string())
            f.write("\n\n")
```
